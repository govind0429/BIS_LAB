```python
import random

traffic = [40, 30, 60, 20]

NUM_PARTICLES = 10
MAX_ITERATIONS = 50
TOTAL_TIME = 120

w = 0.5
c1 = 1.5
c2 = 1.5


def objective(position):
    return sum(v / g for v, g in zip(traffic, position))


def initialize_particle():
    timings = [random.uniform(10, 60) for _ in range(4)]
    total = sum(timings)
    return [g * TOTAL_TIME / total for g in timings]


def pso():
    particles = [initialize_particle()
                 for _ in range(NUM_PARTICLES)]

    velocities = [[0.0] * 4 for _ in range(NUM_PARTICLES)]

    pbest = [p[:] for p in particles]
    pbest_score = [objective(p) for p in particles]

    best_index = pbest_score.index(min(pbest_score))
    gbest = pbest[best_index][:]

    for iteration in range(MAX_ITERATIONS):

        for i in range(NUM_PARTICLES):

            for j in range(4):
                r1 = random.random()
                r2 = random.random()

                velocities[i][j] = (
                    w * velocities[i][j]
                    + c1 * r1 * (pbest[i][j] - particles[i][j])
                    + c2 * r2 * (gbest[j] - particles[i][j])
                )

                particles[i][j] += velocities[i][j]

            particles[i] = [
                max(10, min(60, g)) for g in particles[i]
            ]

            total = sum(particles[i])
            particles[i] = [
                g * TOTAL_TIME / total for g in particles[i]
            ]

            score = objective(particles[i])

            if score < pbest_score[i]:
                pbest[i] = particles[i][:]
                pbest_score[i] = score

        best_index = pbest_score.index(min(pbest_score))
        gbest = pbest[best_index][:]

    return gbest


best_timings = pso()

roads = ["North", "South", "East", "West"]

print("Optimized Traffic Signal Timings:")

for road, timing in zip(roads, best_timings):
    print(f"{road}: {timing:.2f} seconds")

print("Total Green Time:", sum(best_timings))
print("Objective Value:", objective(best_timings))
```
