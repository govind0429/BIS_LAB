```python
import random

# Initialize parameters
population = 10
iterations = 50
a = 2

# Initialize wolf positions
wolves = [random.uniform(1, 100) for i in range(population)]

# Objective function representing a trade-off
# We want to find an optimal 'x' that balances energy cost with performance.
def fitness(x):
    # Energy cost increases with x
    energy = x / 100.0

    # Lifetime and Packet Delivery Ratio (PDR) improve with a higher x
    # Using functions that approach 1 as x increases
    lifetime = 1.0 - (1.0 / (1.0 + 0.05 * x))
    pdr = 1.0 - (1.0 / (1.0 + 0.1 * x))

    # We want to minimize: energy cost, and the penalty for low lifetime/PDR
    # Minimizing (1 - performance) is equivalent to maximizing performance
    return 0.4 * energy + 0.3 * (1.0 - lifetime) + 0.3 * (1.0 - pdr)

# GWO Algorithm
for t in range(iterations):

    wolves.sort(key=fitness)

    alpha = wolves[0]
    beta = wolves[1]
    delta = wolves[2]

    a = 2 - (2 * t / iterations)

    for i in range(population):

        r1 = random.random()
        r2 = random.random()

        A = 2*a*r1 - a
        C = 2*r2

        D1 = abs(C*alpha - wolves[i])
        D2 = abs(C*beta - wolves[i])
        D3 = abs(C*delta - wolves[i])

        X1 = alpha - A*D1
        X2 = beta - A*D2
        X3 = delta - A*D3

        wolves[i] = (X1 + X2 + X3) / 3

        # Restrict to search space [1, 100]
        wolves[i] = max(1, min(100, wolves[i]))

# Display result
best = min(wolves, key=fitness)

print("Optimal Configuration:", best)
print("Minimum Fitness:", fitness(best))
print("Network Energy Optimization Completed")
```
