```python
import random

# Objective function
def fitness(route):
    energy = sum(route)
    return energy

# Initialize nests
n = 10
nests = [[random.randint(1, 10) for _ in range(4)]
         for _ in range(n)]

Pa = 0.25
iterations = 50

best = min(nests, key=fitness)

for i in range(iterations):

    for j in range(n):

        # Generate new solution using random mutation
        new = nests[j].copy()
        k = random.randint(0, 3)
        new[k] = random.randint(1, 10)

        # Select random nest
        r = random.randint(0, n - 1)

        if fitness(new) < fitness(nests[r]):
            nests[r] = new

    # Abandon worst nests
    nests.sort(key=fitness)

    for j in range(int(Pa * n)):
        nests[-(j + 1)] = [
            random.randint(1, 10) for _ in range(4)
        ]

    best = min(nests + [best], key=fitness)

print("Optimal Configuration:", best)
print("Minimum Energy Consumption:", fitness(best))
```
